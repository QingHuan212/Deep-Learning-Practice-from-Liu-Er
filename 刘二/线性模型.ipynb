```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

x_data = [1.0,2.0,3.0]
y_data = [2.0,4.0,6.0]

def predict(x,w,b):
    return x*w+b

def loss(y_hat,y):
    return (y_hat - y) ** 2

w = np.arange(0.0,4.1,0.1)
b = np.arange(-2.0,2.1,0.1)
w,b = np.meshgrid(w,b)

z = np.zeros_like(w)
n = len(x_data)

for i in range(len(w)):
    for j in range(len(w[0])):
        tmpw = w[i][j]
        tmpb = b[i][j]
        for x,y in zip(x_data,y_data):
            z[i][j] += loss(predict(x,tmpw,tmpb),y)

z = z/3.0

fig = plt.figure(figsize=(12,8))
ax = fig.add_subplot(111,projection="3d")

surf = ax.plot_surface(w,b,z, cmap='viridis', edgecolor='none', alpha=0.9)
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=10, label='MSE')

ax.set_xlabel('参数 w')
ax.set_ylabel('参数 b')
ax.set_zlabel('均方误差 (MSE)')
ax.set_title('线性模型 y_hat = w*x + b 的损失曲面')

# 调整视角（方便观察最小值）
ax.view_init(elev=30, azim=45)

plt.show()
```
