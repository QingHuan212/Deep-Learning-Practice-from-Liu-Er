```python
import torch
import numpy as np

x_data = torch.tensor([1.0,2.0,3.0],dtype=torch.float32)
y_data = torch.tensor([2.0,4.0,6.0],dtype=torch.float32)

w = torch.tensor(np.random.normal(0,1,3),requires_grad=True,dtype=torch.float32)
rate = 0.01

def forward(x):
    x_tmp = torch.tensor([x**2,x,1],dtype=torch.float32)
    return torch.dot(w,x_tmp)

def loss(x,y):
    r = forward(x) - y
    return r ** 2

print("before",4,forward(4).data)

for _ in range(10000):
    for x,y in zip(x_data,y_data):
        l = loss(x,y)
        l.backward()
        w.data = w.data - rate*w.grad.data
        w.grad.zero_()


print("last",4,forward(4).data)
```
